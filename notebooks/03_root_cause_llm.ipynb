# Step 3: Root Cause Analysis using LLMs
Here we explore how to pass incident context into a generative LLM (like `llama3`) via Ollama to automatically suggest a root cause and a fix.

```python
import ollama

LLM_MODEL = "llama3.2"

test_log = """
System: auth-service
Error: JWT token validation failed
"""

similar_incident_context = """
Root Cause: Expired secret
Fix: Update token secret
"""

prompt = f"""
You are an expert Site Reliability Engineer.
A new incident has occurred:
{test_log}

Here is a similar historical incident for context:
{similar_incident_context}

Based on this context, what is the most likely root cause for the new incident, and what fix would you propose? Please be concise.
"""
```

```python
print("Analyzing root cause with LLM...")
response = ollama.chat(model=LLM_MODEL, messages=[
  {
    'role': 'user',
    'content': prompt
  }
])

print("\n--- LLM Root Cause Analysis ---")
print(response['message']['content'])
```
