```python
import chromadb
import ollama
import json

CHROMA_PATH = "./chroma_db"
COLLECTION_NAME = "incident_logs"
EMBED_MODEL = "nomic-embed-text"

client = chromadb.PersistentClient(path=CHROMA_PATH)

collection = client.get_or_create_collection(name=COLLECTION_NAME)

fake_logs = [
    {
        "id": "1",
        "system": "payment-service",
        "error": "Database connection timeout after 30 seconds",
        "root_cause": "DB overloaded",
        "fix": "Increase DB pool size"
    },
    {
        "id": "2",
        "system": "auth-service",
        "error": "JWT token validation failed",
        "root_cause": "Expired secret",
        "fix": "Update token secret"
    },
    {
        "id": "3",
        "system": "frontend",
        "error": "API returned 500 Internal Server Error",
        "root_cause": "Null pointer exception",
        "fix": "Add null checks"
    }
]

```

```python
def get_embedding(text):
    response = ollama.embeddings(
        model=EMBED_MODEL,
        prompt=text
    )
    return response["embedding"]

```

```python
for log in fake_logs:

    text = f"""
    System: {log['system']}
    Error: {log['error']}
    Root Cause: {log['root_cause']}
    Fix: {log['fix']}
    """

    embedding = get_embedding(text)

    collection.add(
        ids=[log["id"]],
        embeddings=[embedding],
        documents=[text],
        metadatas=[{"system": log["system"]}]
    )

print("Logs inserted successfully")
```
