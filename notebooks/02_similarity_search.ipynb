# Step 2: Semantic Similarity Search
In this notebook, we'll query the Chroma DB we populated in the first notebook to find similar historical incidents based on a new error log.

```python
import chromadb
import ollama
import json

CHROMA_PATH = "./chroma_db"
COLLECTION_NAME = "incident_logs"
EMBED_MODEL = "nomic-embed-text"

client = chromadb.PersistentClient(path=CHROMA_PATH)
collection = client.get_collection(name=COLLECTION_NAME)
```

```python
def get_embedding(text):
    response = ollama.embeddings(
        model=EMBED_MODEL,
        prompt=text
    )
    return response["embedding"]
```

Let's simulate a new issue occurring in the `payment-service`.

```python
new_incident = """
System: payment-service
Error: DB Connection timed out while trying to process payment
"""

print("Searching for similar past incidents...")
query_embedding = get_embedding(new_incident)

results = collection.query(
    query_embeddings=[query_embedding],
    n_results=1
)

print("\n--- Top Match ---")
for doc in results['documents'][0]:
    print(doc)
```
